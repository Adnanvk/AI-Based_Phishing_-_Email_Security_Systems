```python
from yfiles_jupyter_graphs import GraphWidget

def build_graph(nodes, edges):
    w = GraphWidget()
    w.directed = True
    w.nodes = [
        {"id": n["id"], "properties": {
            "label":       n["label"],
            "cat":         n["cat"],
            "description": n["description"],
            "url":         n.get("url",""),
        }} for n in nodes
    ]
    w.edges = [
        {"id": e["id"], "start": e["s"], "end": e["e"],
         "properties": {"label": e["l"]}} for e in edges
    ]
    w.set_node_color_mapping(node_color)
    w.set_node_scale_factor_mapping(node_scale)
    w.set_edge_color_mapping(edge_color)
    return w

w_full = build_graph(NODES, EDGES)
print(f"Full graph: {len(NODES)} nodes · {len(EDGES)} edges")
print("Interact: click node for details · drag to move · scroll/pinch to zoom")
display(w_full)
```

## 🔍 Step 5 — Focused Subgraph Explorer

Each cell below renders a filtered view. Run only the cells you need.

```python
# ── HELPER FUNCTIONS ─────────────────────────────────────────────────────────
def subgraph_by_cat(cats, title=""):
    ids = {n["id"] for n in NODES if n["cat"] in cats}
    fn = [n for n in NODES if n["id"] in ids]
    fe = [e for e in EDGES if e["s"] in ids and e["e"] in ids]
    print(f"  {title or str(cats)}: {len(fn)} nodes, {len(fe)} edges")
    display(build_graph(fn, fe))

def subgraph_ego(seed_ids, depth=1, title=""):
    ids = set(seed_ids)
    for _ in range(depth):
        new = set()
        for e in EDGES:
            if e["s"] in ids: new.add(e["e"])
            if e["e"] in ids: new.add(e["s"])
        ids |= new
    fn = [n for n in NODES if n["id"] in ids]
    fe = [e for e in EDGES if e["s"] in ids and e["e"] in ids]
    print(f"  {title}: {len(fn)} nodes, {len(fe)} edges")
    display(build_graph(fn, fe))

print("Helpers ready.")
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# Build NetworkX directed graph
G = nx.DiGraph()
nlu = {n["id"]: n for n in NODES}
for n in NODES:
    G.add_node(n["id"], **n)
for e in EDGES:
    G.add_edge(e["s"], e["e"], label=e["l"])

# Degree centrality (total connections)
deg = dict(G.degree())
top20 = sorted(deg.items(), key=lambda x: x[1], reverse=True)[:20]

lids    = [x[0] for x in top20]
vals    = [x[1] for x in top20]
cols    = [CAT_COLORS.get(nlu[l]["cat"], "#CBD5E1") for l in lids]
labels  = [nlu[l]["label"].replace("\n"," ") for l in lids]

fig, axes = plt.subplots(1, 2, figsize=(18, 8))
fig.patch.set_facecolor("#0D1117")

# ── Left: centrality bar chart ──────────────────────────────────────────────
ax = axes[0]
ax.set_facecolor("#161B22")
bars = ax.barh(range(len(lids)), vals, color=cols, edgecolor="#30363D", linewidth=0.6)
ax.set_yticks(range(len(lids)))
ax.set_yticklabels(labels, color="#C9D1D9", fontsize=8.5)
ax.set_xlabel("Degree Centrality (total connections)", color="#8B949E", fontsize=10)
ax.set_title("Top 20 Most Connected Nodes\n(Compliance Pinch-Points)", color="#F0F6FC",
             fontsize=12, fontweight="bold", pad=12)
ax.tick_params(colors="#8B949E")
for sp in ["top","right"]: ax.spines[sp].set_visible(False)
for sp in ["bottom","left"]: ax.spines[sp].set_color("#30363D")
ax.invert_yaxis()
for bar, val in zip(bars, vals):
    ax.text(bar.get_width()+0.15, bar.get_y()+bar.get_height()/2,
            str(val), va="center", color="#8B949E", fontsize=8)

# ── Right: risk level heat map ──────────────────────────────────────────────
ax2 = axes[1]
ax2.set_facecolor("#161B22")
risk_data = [
    ("R-01 False Negative",        "Med",    "Critical", "HIGH",     3, 5),
    ("R-02 False Positive",        "Med-Hi", "High",     "HIGH",     4, 4),
    ("R-03 Art.22 Non-Compliance", "High",   "Medium",   "HIGH",     5, 3),
    ("R-04 Algorithmic Bias",      "Med",    "High",     "MEDIUM",   3, 4),
    ("R-05 Automation Bias",       "High",   "Medium",   "MED-HIGH", 5, 3),
    ("R-06 Adversarial ML",        "Low-Med","Critical", "MEDIUM",   2, 5),
    ("R-07 DPIA Absence",          "High",   "Medium",   "MED-HIGH", 5, 3),
]
heat_colors = {"HIGH":"#DC2626","MED-HIGH":"#EA580C","MEDIUM":"#D97706","LOW":"#16A34A"}
y_pos = range(len(risk_data))
for i, (name, like, imp, level, lv, iv) in enumerate(risk_data):
    col = heat_colors.get(level, "#6B7280")
    ax2.barh(i, lv, color=col, alpha=0.3, height=0.4)
    ax2.barh(i, iv, left=5, color=col, alpha=0.6, height=0.4)
    ax2.text(-0.1, i, name, ha="right", va="center", color="#C9D1D9", fontsize=8.5)
    ax2.text(lv/2, i, like, ha="center", va="center", color="#F0F6FC", fontsize=7, fontweight="bold")
    ax2.text(5+iv/2, i, imp, ha="center", va="center", color="#F0F6FC", fontsize=7, fontweight="bold")
    ax2.text(10.3, i, level, ha="left", va="center",
             color=col, fontsize=8.5, fontweight="bold")

ax2.set_yticks([])
ax2.set_xticks([0,2.5,5,7.5,10])
ax2.set_xticklabels(["0","L'hood →","5","Impact →","10"], color="#8B949E", fontsize=8)
ax2.set_xlim(-5.5, 13)
ax2.set_title("Risk Register Heat Map\n(Likelihood × Impact × Regulatory Level)", color="#F0F6FC",
              fontsize=12, fontweight="bold", pad=12)
for sp in ["top","right","left","bottom"]: ax2.spines[sp].set_color("#30363D")
ax2.axvline(5, color="#30363D", linewidth=1, linestyle="--")

# Legend
patches = [mpatches.Patch(color=c, label=k) for k, c in CAT_COLORS.items()]
fig.legend(handles=patches, loc="lower center", ncol=5, facecolor="#161B22",
           edgecolor="#30363D", labelcolor="#C9D1D9", fontsize=7.5,
           bbox_to_anchor=(0.5, -0.02))

plt.tight_layout(rect=[0,0.06,1,1])
plt.savefig("compliance_analysis.png", dpi=150, bbox_inches="tight", facecolor="#0D1117")
plt.show()

print("\nKEY INSIGHTS:")
print(f"  Most connected node: {nlu[top20[0][0]]['label'].replace(chr(10),' ')} ({top20[0][1]} connections)")
print(f"  GDPR Art.22 connections: {deg.get('GDPR22', 0)}")
print(f"  HITL Artefact connections: {deg.get('ART_HITL', 0)}")
```
